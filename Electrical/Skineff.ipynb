```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import jv

# Spatial domain
x = np.linspace(0, 0.0098, 500)

epsilon0=8.854187817e-12  # Vacuum permittivity units F/m
mu0=4e-7*np.pi          # Vacuum permeability units H/m
sigmacopper=5.96e7  # Conductivity of copper S/m
f=600
w=2*np.pi*f  # Angular frequency for 1 GHz
# Complex wave number
k = w*np.sqrt(epsilon0*mu0*(-1 + 1j*sigmacopper/(epsilon0*w)))

# Bessel function J0(kx)
J = jv(0, k*x)

# Plot real and imaginary parts
plt.figure()
plt.plot(x, J.real, label="Re[J₀(kx)]",color="blue")
plt.plot(x, J.imag, label="Im[J₀(kx)]",color="red")
plt.plot(x, np.abs(J), label="|J₀(kx)|", linestyle='--', color="black")
plt.xlabel("x")
plt.ylabel("J₀(kx)")
plt.legend()
plt.show()
```
