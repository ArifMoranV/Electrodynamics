```python
import numpy as np
xs = np.linspace(0, 1/60, 1000)
w= 2 * np.pi *60
Zn=0.01

phi=np.array([0, -2*np.pi/3, 2*np.pi/3,0])
M12=0.15e-6
Mac=0.075e-6
L=0.5e-6
XL=1j*w*L
XM12=1j*w*M12
XMac=1j*w*Mac
R=240
ZLA=R+ XL
ZLB=R+ XL
ZLC=R+ XL
V=120
M= np.array([[ZLA, XM12, XMac,Zn],
             [XM12, ZLB, XM12, Zn],
             [XMac, XM12, ZLC, Zn],
             [1, 1, 1, -1]])

phse=np.exp(1j * phi)
r = V * phse
r[-1] = 0
x = np.linalg.solve(M, r)
print(x)
```

```python
import matplotlib.pyplot as plt
i0=(x)
plt.plot(xs, np.imag(i0[0]*np.exp(1j*w*xs)), label='Phase A Current')
plt.plot(xs, np.imag(i0[1]*np.exp(1j*w*xs)), label='Phase B Current')
plt.plot(xs, np.imag(i0[2]*np.exp(1j*w*xs)), label='Phase C Current')
plt.plot(xs, np.imag(i0[3]*np.exp(1j*w*xs)), label='Neutral Current')
#plt.plot(xs, np.imag(np.sum(i0[:-1])*np.exp(1j*w*xs)), label='Neutral Current')
plt.legend()
```
